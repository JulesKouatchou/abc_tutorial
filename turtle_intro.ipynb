# Introduction to Turtle in Colab

## What is `turtle`?

- Python library that provides a virtual drawing board where you can control a cursor to create shapes, patterns, and digital art using simple code.
- Popular for introducing kids and beginners to core computer programming concepts.
- Learners get immediate visual results from their text commands.

## Key Commands Cheat Sheet

| Command | Description | Example |
| --- | --- | --- |
| `initializeTurtle()` | Sets up or resets a fresh drawing canvas | `initializeTurtle()` |
| `forward(pixels)/fd(pixels)` | Moves the turtle straight forward | `fd(50)` |
| `backward(pixels)/bk(pixels)` | Moves the turtle backward | `bk(50)` |
| `right(angle)/rt(angle)` | Turns the turtle clockwise | `rt(90)` |
| `left(angle)/lt(angle)` | Turns the turtle counter-clockwise | `lt(45)` |
| `color('color_name')` | Changes the pen line color | `color('red')` |
| `penup()/pu()` | Lifts pen so you can move without drawing | `pu()` |
| `pendown()/pd()` | Puts pen down to start drawing again | `pd()` |
| `goto(x, y)` | Moves turtle directly to a specific pixel coordinate | `goto(100, 200)` |

## First step

- Before you start drawing anything, you first need to clear the canvas using the call: `initializeTurtle()`.
- Every time we have a new drawing, we will start with that call.

````python
initializeTurtle()
````

## Speed of the drawing

- When you draw an object, you have control of how fast you want to do it.
- The `speed(number)` function determine the speed where `number` range from `1` (slow) to `13` (fast).

## Simple drawings

### Line

````python
initializeTurtle()   # Clears canvas for a clean start
speed(1)             # Set speed to 1
forward(100)         # Move forward 100 pixels
````

````python
initializeTurtle()   # Clears canvas for a clean start
color('red')         # Change the color to red
speed(1)             # Set speed to 1
forward(100)         # Move forward 100 pixels
````

### Two connected lines

We want to draw a line, rotate 90 degrees and draw another line.

````python
initializeTurtle() # Clears canvas for a clean start
color('blue')      # Change pen color to blue
speed(1)           # Set speed to 1

forward(100)       # Move forward 100 pixels
right(90)          # Turn right 90 degrees
forward(150)       # Move forward 150 pixels
````

### Draw a staircase with two steps

````python
initializeTurtle() # Clears canvas for a clean start
color('blue')      # Change pen color to blue
speed(1)           # Set speed to 1

fd(100)            # Move forward 100 pixels
rt(90)             # Turn right 90 degrees
fd(150)            # Move forward 150 pixels
lt(90)             # Turn left 90 degrees
fd(100)            # Move forward 100 pixels
rt(90)             # Turn right 90 degrees
fd(150)            # Move forward 150 pixels
````

### Draw a square

````python
initializeTurtle() # Clears canvas for a clean start
color('blue')      # Change pen color to blue
speed(1)           # Set speed to 1

fd(100)            # Move forward 100 pixels
rt(90)             # Turn right 90 degrees
fd(100)            # Move forward 150 pixels
rt(90)             # Turn right 90 degrees
fd(100)            # Move forward 100 pixels
rt(90)             # Turn right 90 degrees
fd(100)            # Move forward 150 pixels
````

## Concept of a loop

- In the above example, it appears that we perform the same operation (drawing a line and rotating 90 degrees)four times.
- We want to simplify the code by introducing a loop.
- A __loop__ is a programming construct that allows you to __repeat a set of instructions multiple times__.

In Python, the basic syntax to write a look is:

```python
for i in range(n):
    <statements>
```

Here `n` is the number of times the `<statements>` is repeated.

````python
for i in range(4):
    print(i)
````

````python
for i in range(3):
    print('Hello learners of Turtle!')
````

### Drawing a square using a loop

````python
initializeTurtle() # Clears canvas for a clean start
color('blue')      # Change pen color to blue
speed(5)           # Set speed between 1 (slow) and 13 (fast)

for i in range(4):
    forward(100)   # Move forward 100 pixels
    right(90)
````

### Drawing a star

````python
initializeTurtle()
color('orange')
speed(7)

for i in range(5):
    forward(150)
    rt(144)    # Turn 144 degrees for a standard 5-point star
````

### Discontinuous line

````python
initializeTurtle()
right(90)
for i in range(10):
    forward(10)
    penup()
    forward(5)
    pendown()
````
